```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load Dataset
df = pd.read_csv("C:/Users/shrav/OneDrive/Desktop/Rani/Car Market Trends Analysis with Car Dekho Data.csv")
```

```python
# Display first 5 rows
print(df.head())
```

```python
# Check dataset information
print(df.info())
```

```python
# 1. Average Selling Price by Fuel Type
fuel_avg = df.groupby("Fuel_Type")["Selling_Price"].mean()

plt.figure(figsize=(7, 5))
fuel_avg.plot(kind="bar")

plt.title("Average Selling Price by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Average Selling Price")
plt.xticks(rotation=0)

plt.show()
```

```python
# 2. Average Selling Price by Seller Type
seller_avg = df.groupby("Seller_Type")["Selling_Price"].mean()

plt.figure(figsize=(7, 5))
seller_avg.plot(kind="bar")

plt.title("Average Selling Price by Seller Type")
plt.xlabel("Seller Type")
plt.ylabel("Average Selling Price")
plt.xticks(rotation=0)

plt.show()
```

```python
# 3. Average Selling Price by Transmission
transmission_avg = df.groupby("Transmission")["Selling_Price"].mean()

plt.figure(figsize=(7, 5))
transmission_avg.plot(kind="bar")

plt.title("Average Selling Price by Transmission")
plt.xlabel("Transmission")
plt.ylabel("Average Selling Price")
plt.xticks(rotation=0)

plt.show()
```

```python
# 4. Present Price vs Selling Price
plt.figure(figsize=(7, 5))

plt.scatter(
    df["Present_Price"],
    df["Selling_Price"],
    alpha=0.6
)

plt.title("Present Price vs Selling Price")
plt.xlabel("Present Price")
plt.ylabel("Selling Price")

plt.show()
```
